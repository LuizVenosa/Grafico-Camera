```python
unique_deputados_uri = votos['deputado_uri'].unique()
votos_u = votos[votos['deputado_uri'].isin(unique_deputados_uri)].drop_duplicates(subset='deputado_uri')
```

```python
votos_u
```

```python
# Create sets of names for faster lookup
centrao_nomes = set(centrao['0'])
direita_nomes = set(direita['0'])
esquerda_nomes = set(esquerda['0'])

# Define a function to determine the alignment
def get_alinhamento(nome):
    if nome in centrao_nomes:
        return 'centrao'
    elif nome in direita_nomes:
        return 'direita'
    elif nome in esquerda_nomes:
        return 'esquerda'
    else:
        return 'outros' # Or pd.NA if you prefer missing values

# Apply the function to create the 'alinhamento' column
votos_u['alinhamento'] = votos_u['deputado_nome'].apply(get_alinhamento)

# Display the first few rows with the new column
print(votos_u[['deputado_nome', 'alinhamento']].head())
```

```python
essencial_autor = votos_u[['deputado_id', 'deputado_uri', 'deputado_nome', 'alinhamento']]
```

```python
import matplotlib.pyplot as plt

# Calculate the counts and get the top 10 themes
top_temas = temas_autor_final.groupby('tema')['uriProposicao'].count().nlargest(10)

# Sort ascending for horizontal bar plot (largest bar appears at the top)
top_temas_sorted = top_temas.sort_values(ascending=True)

# Create the plot
plt.figure(figsize=(12, 7)) # Adjust figure size for better readability
top_temas_sorted.plot(kind='barh', color='skyblue') # Use horizontal bars and a specific color

# Add title and labels with adjusted font sizes
plt.title('Top 10 Temas', fontsize=16)
plt.xlabel('Number of Propositions', fontsize=12)
plt.ylabel('Tema', fontsize=12)

# Adjust tick label sizes
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Add grid lines for easier comparison
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Ensure layout is tight to prevent labels overlapping
plt.tight_layout()



# Display the plot
plt.show()
```

```python
# Calculate the counts and get the bottom 10 themes
bottom_temas = temas_autor_final.groupby('tema')['uriProposicao'].count().nsmallest(10)

# Sort ascending for horizontal bar plot (smallest bar appears at the top)
# nsmallest already returns sorted data, but explicitly sorting for clarity
bottom_temas_sorted = bottom_temas.sort_values(ascending=True)

# Create the plot
plt.figure(figsize=(12, 7)) # Adjust figure size for better readability
bottom_temas_sorted.plot(kind='barh', color='lightcoral') # Use horizontal bars and a different color

# Add title and labels with adjusted font sizes
plt.title('Bottom 10 Temas', fontsize=16)
plt.xlabel('Number of Propositions', fontsize=12)
plt.ylabel('Tema', fontsize=12)

# Adjust tick label sizes
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Add grid lines for easier comparison
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Ensure layout is tight to prevent labels overlapping
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Calculate the normalized value counts for the 'tema' column
tema_distribution = temas_autor_final['tema'].value_counts(normalize=True)

# Sort the values in ascending order for the horizontal bar plot
tema_distribution_sorted = tema_distribution.sort_values(ascending=True)

# Create the plot
# Increased height for better readability with potentially many themes
plt.figure(figsize=(12, 15))
# Use horizontal bars and a specific color
tema_distribution_sorted.plot(kind='barh', color='mediumpurple')

# Add title and labels with adjusted font sizes
plt.title('Distribution of Temas (Normalized)', fontsize=16)
plt.xlabel('Proportion of Propositions', fontsize=12)
plt.ylabel('Tema', fontsize=12)

# Adjust tick label sizes
plt.xticks(fontsize=10)
# Smaller font size for y-axis if many themes
plt.yticks(fontsize=8)

# Add grid lines for easier comparison
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Ensure layout is tight to prevent labels overlapping
plt.tight_layout()

# Display the plot
plt.show()
```
